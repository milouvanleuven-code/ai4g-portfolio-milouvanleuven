## Learning Wave 1 — The String Toolkit

### 🧠 First: what *is* a string, really?

A **string** is just a **row of characters** — exactly like a list, but every slot holds one character. So everything you already know about lists still works:

- `text[0]` is the first character, `text[-1]` the last
- `len(text)` is how many characters there are
- `"bike" in text` checks whether that word appears

What strings add is a **toolbox of methods** for cleaning and reshaping text. The ones you will use constantly:

| Method | What it does |
|---|---|
| `.strip()` | remove spaces / newlines at the start and end |
| `.lower()` / `.upper()` | change case, so `"Bike"` and `"bike"` count as the same word |
| `.replace(old, new)` | swap one piece of text for another |
| `.split()` | cut a sentence into a **list of words** |
| `" ".join(words)` | glue a list of words back into one sentence |
| `.count("x")` | how many times `"x"` appears |

**The big idea for the whole week:** a computer cannot understand a sentence. The first move in *every* text and AI pipeline is to **clean** the text and **split it into pieces you can count.**

### Exercise 1: Clean one piece of feedback
⏱ **15 minutes** | Slide 7

**Context:** A city collects open-text feedback about a neighbourhood. Raw text is messy — your first job is to clean a single response.

**Steps:**
1. Create `raw = "  The NEW bike lanes are GREAT!! but the lights are too short.  "`.
2. Print it with `repr()` so you can see the spaces, then `strip()` it into `clean`.
3. Make a `lower()`-case version.
4. How many **characters** does `clean` have? How many **words**? (`len()` and `.split()`)
5. Does the feedback mention `"bike"`? And `"lights"`? Use the `in` operator on the lowercase version.
6. `replace()` the `"!!"` with `"."` and print the final, clean sentence.

````python
# Exercise 1: Clean one piece of feedback — SOLUTION
raw = "  The NEW bike lanes are GREAT!! but the lights are too short.  "

print("Raw:     ", repr(raw))
clean = raw.strip()
print("Stripped:", repr(clean))

lower = clean.lower()
print("Lowercase:", lower)

print("Characters:", len(clean))
print("Words:     ", len(clean.split()))

print("Mentions 'bike'?  ", "bike" in lower)
print("Mentions 'lights'?", "lights" in lower)

final = clean.replace("!!", ".")
print("Final:", final)
````

## Learning Wave 2 — From Text to Numbers

One response is easy. But a city gets *thousands*. To find what people care about, you count words across **all** of them — this is called a **bag of words**, and it is the foundation of almost all classic NLP.

The recipe reuses exactly what you already know:
1. `for` over every response (Week 2)
2. `.lower().split()` each response into words (Week 4)
3. the **counting pattern** from Week 3: `counts[word] = counts.get(word, 0) + 1`

### Exercise 2: Which words come up most?
⏱ **15 minutes** | Slide 10

**Given:**
```python
responses = [
    "The bike lanes are great but too narrow",
    "More bike parking please the parking is always full",
    "Traffic lights are too short for pedestrians",
    "Great new park but we need more lights at night",
    "The park is lovely and the bike lanes are a great idea",
]
```

**Steps:**
1. Build a dictionary `counts` of how often each (lowercased) word appears, using the counting pattern.
2. Print every word with its count using `.items()`.
3. Find the **most mentioned** word with the comparison-variable pattern (Weeks 2–3 — no need for `max()` on a dict).
4. **Bonus:** make a `stopwords` list (`"the"`, `"are"`, `"but"`, `"and"`, `"a"`, `"is"`, `"for"`, …) and skip those words with `continue`, so only meaningful words remain.

````python
# Exercise 2: Which words come up most? — SOLUTION
responses = [
    "The bike lanes are great but too narrow",
    "More bike parking please the parking is always full",
    "Traffic lights are too short for pedestrians",
    "Great new park but we need more lights at night",
    "The park is lovely and the bike lanes are a great idea",
]

# 1. Count every word (counting pattern from Week 3)
counts = {}
for response in responses:
    for word in response.lower().split():
        counts[word] = counts.get(word, 0) + 1

# 2. Show each word and its count
for word, n in counts.items():
    print(word, ":", n)

# 3. Most mentioned word (comparison-variable pattern)
top_word = ""
top_count = 0
for word, n in counts.items():
    if n > top_count:
        top_count = n
        top_word = word
print("\nMost mentioned:", top_word, "(" + str(top_count) + "x)")

# 4. Bonus: skip common stopwords
stopwords = ["the", "are", "but", "and", "a", "is", "for", "we", "at", "to", "more", "too"]
meaningful = {}
for word, n in counts.items():
    if word in stopwords:
        continue
    meaningful[word] = n
print("Meaningful words only:", meaningful)
````

## Learning Wave 3 — Reading & Writing Files

So far the data lived *inside* your code. Real data lives in **files**. Python opens files with `open()`, and the safe way is the `with` block — it closes the file for you automatically:

```python
# Writing
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")      # \n = a new line

# Reading, line by line
with open("notes.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())      # strip() removes the trailing newline
```

- `"w"` = write (overwrites the whole file), `"a"` = append, `"r"` = read.
- Always pass `encoding="utf-8"` so accents and symbols don't break.

### Exercise 3: Read feedback from a file
⏱ **20 minutes** (incl. reflection) | Slide 13

**Steps:**
1. Write the lines below to a file `feedback.txt` (use `writelines()`; each line already ends in `\n`).
2. Open `feedback.txt` and loop over it line by line. `strip()` each line; if the line is empty, `continue`.
3. Count the **total** responses, and count how many contain the word `"ALERT"`. Print each urgent line.
4. Write a short report to `report.txt`: total responses and number of urgent ones.
5. **Reflection** (markdown cell below).

```python
lines = [
    "The bike lanes are great\n",
    "ALERT broken glass on the path near the school\n",
    "More lights please it is too dark at night\n",
    "\n",
    "ALERT the crossing feels unsafe for children\n",
    "Lovely new park thank you\n",
]
```

````python
# Exercise 3: Read feedback from a file — SOLUTION
# 1. Write the data file (in real life this file already exists)
lines = [
    "The bike lanes are great\n",
    "ALERT broken glass on the path near the school\n",
    "More lights please it is too dark at night\n",
    "\n",
    "ALERT the crossing feels unsafe for children\n",
    "Lovely new park thank you\n",
]
with open("feedback.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)

# 2 & 3. Read it back, skip empty lines, count and flag urgent ones
total = 0
urgent = 0
with open("feedback.txt", "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line == "":
            continue
        total = total + 1
        if "ALERT" in line:
            urgent = urgent + 1
            print("URGENT:", line)

print("Total responses:", total)
print("Urgent responses:", urgent)

# 4. Write a small report
with open("report.txt", "w", encoding="utf-8") as f:
    f.write("Feedback report\n")
    f.write("Total responses: " + str(total) + "\n")
    f.write("Urgent responses: " + str(urgent) + "\n")
print("Report written to report.txt")
````

### 📊 Community Feedback Analyzer

**Goal:** Build a program that **reads a file of citizen feedback, analyses the text, and writes a report** — a tool a city or NGO could actually use to *listen at scale*.

> **This assignment is more open than before — there is no step-by-step recipe.** You decide how to structure your program (functions, a menu, or one pipeline). What matters is that your solution **uses what you learned today**, end to end.

#### ✅ Your program **must** use all of these (this week's skills):
- **Read** the feedback from a file with a `with open(...)` block
- **Clean** the text with string methods (`strip()`, `lower()`, `replace()` and/or `.strip(".,!?")`)
- **Split** responses into words with `.split()`
- **Count** word frequencies with a dictionary and the counting pattern
- **Write** a readable report to a new file
- (Reuse freely from Weeks 1–3: `if/else`, loops, functions, lists, dicts.)

#### 📋 What the report must contain (the *what* — the *how* is yours):
1. How many responses were analysed, and the **average number of words** per response.
2. The **top 5 keywords** (most frequent words, ignoring stopwords).
3. The number of **urgent** responses (containing words like `unsafe`, `broken`, `alert`, `danger`) — and list them.
4. Save the whole report to `feedback_report.txt` **and** print it to the screen.

#### 🌟 Bonus (for fast finishers):
- A tiny **sentiment score**: count words from a `positive` list vs. a `negative` list and report the mood.
- Let the user **search** for a word and report how many responses mention it.

#### 💡 Tips:

2. A `KeyError` means you reached for a key that isn't there — reach for `.get()` instead.
3. Write one small helper function per job (`read_responses`, `count_words`, …). It makes the program readable and is exactly how real text pipelines are built.

#### ✍️ Reflection (markdown cell below the solution).

````python
# Big Assignment: Community Feedback Analyzer — REFERENCE SOLUTION
# This is ONE way to solve it. Students may structure their program differently
# as long as it reads a file, cleans + splits + counts text, and writes a report.

# --- Setup: create a feedback file to analyse (normally this already exists) ---
feedback_lines = [
    "The new bike lanes are great but they are too narrow\n",
    "Please add more lights the park is too dark at night\n",
    "ALERT broken glass near the school entrance\n",
    "The park is lovely thank you for the new benches\n",
    "Traffic is unsafe for children crossing to school\n",
    "More bike parking please it is always full\n",
    "ALERT the footbridge feels unsafe when it rains\n",
    "Great idea the lanes make cycling safer\n",
]
with open("community_feedback.txt", "w", encoding="utf-8") as f:
    f.writelines(feedback_lines)

STOPWORDS = ["the", "a", "an", "are", "is", "it", "they", "but", "and", "to",
             "for", "of", "at", "in", "more", "please", "when", "make", "too"]
URGENT_WORDS = ["alert", "unsafe", "broken", "danger", "dark"]


def read_responses(filename):
    responses = []
    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line != "":
                responses.append(line)
    return responses


def count_words(responses):
    counts = {}
    for response in responses:
        for word in response.lower().split():
            word = word.strip(".,!?")
            if word == "" or word in STOPWORDS:
                continue
            counts[word] = counts.get(word, 0) + 1
    return counts


def top_words(counts, n):
    counts = dict(counts)          # work on a copy
    result = []
    for i in range(n):
        best_word = ""
        best_count = 0
        for word, c in counts.items():
            if c > best_count:
                best_count = c
                best_word = word
        if best_word == "":
            break
        result.append((best_word, best_count))
        counts[best_word] = 0      # so it is not picked again
    return result


def urgent_responses(responses):
    flagged = []
    for response in responses:
        for word in URGENT_WORDS:
            if word in response.lower():
                flagged.append(response)
                break
    return flagged


# --- Run the pipeline ---
responses = read_responses("community_feedback.txt")
counts = count_words(responses)
total_words = sum(counts.values())
flagged = urgent_responses(responses)

report = []
report.append("=== Community Feedback Report ===")
report.append("Responses analysed: " + str(len(responses)))
report.append("Average words per response: " + str(round(total_words / len(responses), 1)))
report.append("")
report.append("Top 5 keywords:")
for word, n in top_words(counts, 5):
    report.append("  " + word + ": " + str(n))
report.append("")
report.append("Urgent responses (" + str(len(flagged)) + "):")
for response in flagged:
    report.append("  - " + response)

report_text = "\n".join(report)
print(report_text)

with open("feedback_report.txt", "w", encoding="utf-8") as f:
    f.write(report_text + "\n")
print("\nReport saved to feedback_report.txt")
````

#### 🌍 Bonus Reflection
**Your analyzer decides what counts as a "keyword" and what counts as "urgent". Those choices shape what the city sees first. Who should decide which words are treated as urgent — and what gets missed when a machine ranks feedback by word frequency?**

> **Sample answer (instructor):** Frequency rewards whatever is easy to say in common words; a rare but serious report ("the lift has been broken for a month") can be drowned out by many small ones ("nice park"). The `URGENT_WORDS` list is a value judgement coded by the programmer — miss a word like "scared" or a language other than English and that feedback silently drops in priority. A useful tool keeps a human in the loop, makes its keyword lists visible and editable, and treats the ranking as a *starting point for attention*, not an automatic decision. This is the same lesson as the food-bank allocation in Week 3: the moment you sort or filter people's input, you have made a policy choice.

## 5 Things to Remember

1. **A string is a row of characters** — indexing, slicing, `len()` and `in` work just like with lists.
2. **Clean before you count** — `.strip()`, `.lower()` and `.replace()` turn messy text into something comparable (`"Bike!"` and `"bike"` become the same).
3. **`.split()` turns a sentence into a list of words; `" ".join()` turns it back** — this is *tokenisation*, the first step of every NLP system.
4. **Counting words = the Week 3 counting pattern** — `counts[word] = counts.get(word, 0) + 1` is a "bag of words".
5. **Files in, results out** — `with open(...) as f:` reads real data and writes reports safely. And remember: *the words missing from the file are missing from everything you build on it.*
